```python
iris_url = "https://archive.ics.uci.edu/ml/machine-learning-databases/iris/iris.data"
print(repr(iris_url))
```

```python

import numpy as np
import urllib.request
import ssl
import certifi
import io

context = ssl.create_default_context(cafile=certifi.where())

with urllib.request.urlopen(iris_url, context=context) as response:
    iris_text = response.read().decode("utf-8")

iris_data = np.genfromtxt(
    io.StringIO(iris_text),
    delimiter=",",
    dtype=str
)

print(iris_data[:5])
```

```python
matches = iris_data[iris_data[:, 3].astype(float) > 1]
index = np.where(iris_data[:, 3].astype(float) > 1)[0][0]
first_match = matches[0]
print(first_match)
print(index)
```
